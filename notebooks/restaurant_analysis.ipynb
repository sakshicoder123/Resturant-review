## 4. Data Preprocessing

```python
# Columns to DROP
drop_columns = [
    'Restaurant ID',       # Unique ID
    'Restaurant Name',     # Text identifier
    'Address',             # Free text
    'Locality',            # Redundant with City
    'Locality Verbose',    # Redundant
    'Currency',            # Redundant with Country Code
    'Rating color',        # DATA LEAKAGE
    'Rating text',         # DATA LEAKAGE
    'Switch to order menu',# Nearly constant
    'Country Code',        # Using City instead
]

print('Columns dropped:')
for col in drop_columns:
    reason = 'DATA LEAKAGE' if col in ['Rating color', 'Rating text'] else 'Not useful for prediction'
    print(f'  - {col}: {reason}')
```

```python
# Feature Engineering
# Num_Cuisines
df['Num_Cuisines'] = df['Cuisines'].fillna('Unknown').apply(lambda x: len(str(x).split(',')))

# Primary Cuisine
df['Primary_Cuisine'] = df['Cuisines'].fillna('Unknown').apply(lambda x: str(x).split(',')[0].strip())
top_cuisines = df['Primary_Cuisine'].value_counts().head(20).index.tolist()
df['Primary_Cuisine'] = df['Primary_Cuisine'].apply(lambda x: x if x in top_cuisines else 'Other')

# Binary encoding
binary_cols = ['Has Table booking', 'Has Online delivery', 'Is delivering now']
for col in binary_cols:
    df[col + '_enc'] = (df[col] == 'Yes').astype(int)

# Handle missing cuisines
df['Cuisines'] = df['Cuisines'].fillna('Unknown')

# Reduce City cardinality
top_cities = df['City'].value_counts().head(30).index.tolist()
df['City'] = df['City'].apply(lambda x: x if x in top_cities else 'Other')

print(f'Num_Cuisines range: {df["Num_Cuisines"].min()} - {df["Num_Cuisines"].max()}')
print(f'Primary Cuisine categories: {df["Primary_Cuisine"].nunique()}')
print(f'City categories: {df["City"].nunique()}')
```

```python
# Define features and target
TARGET = 'Aggregate rating'
y = df[TARGET].values

feature_columns_numeric = [
    'Average Cost for two', 'Longitude', 'Latitude', 'Votes',
    'Price range', 'Num_Cuisines',
    'Has Table booking_enc', 'Has Online delivery_enc', 'Is delivering now_enc',
]

feature_columns_categorical = ['City', 'Primary_Cuisine']

X = df[feature_columns_numeric + feature_columns_categorical]

print(f'Features: {len(feature_columns_numeric)} numerical + {len(feature_columns_categorical)} categorical')
print(f'Target: {TARGET}')
print(f'X shape: {X.shape}')
print(f'y shape: {y.shape}')
```

```python
# Preprocessing pipeline
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), feature_columns_numeric),
        ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False, drop='first'),
         feature_columns_categorical),
    ]
)

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print(f'Training set: {X_train.shape[0]} samples ({X_train.shape[0]/len(X)*100:.1f}%)')
print(f'Testing set:  {X_test.shape[0]} samples ({X_test.shape[0]/len(X)*100:.1f}%)')
```

## 5. Model Training & Evaluation

```python
def evaluate_model(name, y_true, y_pred):
    """Calculate all evaluation metrics."""
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, y_pred)
    return {'Model': name, 'MAE': mae, 'MSE': mse, 'RMSE': rmse, 'R² Score': r2}

results = []
```

```python
# Model 1: Linear Regression
print('=== Linear Regression ===')
lr_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('regressor', LinearRegression())
])
lr_pipeline.fit(X_train, y_train)
lr_pred = lr_pipeline.predict(X_test)
lr_metrics = evaluate_model('Linear Regression', y_test, lr_pred)
results.append(lr_metrics)
for k, v in lr_metrics.items():
    if k != 'Model': print(f'  {k}: {v:.4f}')
```

```python
# Model 2: Decision Tree
print('=== Decision Tree Regressor ===')
dt_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('regressor', DecisionTreeRegressor(random_state=42))
])
dt_pipeline.fit(X_train, y_train)
dt_pred = dt_pipeline.predict(X_test)
dt_metrics = evaluate_model('Decision Tree', y_test, dt_pred)
results.append(dt_metrics)
for k, v in dt_metrics.items():
    if k != 'Model': print(f'  {k}: {v:.4f}')
```

```python
# Model 3: Random Forest
print('=== Random Forest Regressor ===')
rf_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('regressor', RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1))
])
rf_pipeline.fit(X_train, y_train)
rf_pred = rf_pipeline.predict(X_test)
rf_metrics = evaluate_model('Random Forest', y_test, rf_pred)
results.append(rf_metrics)
for k, v in rf_metrics.items():
    if k != 'Model': print(f'  {k}: {v:.4f}')
```

```python
# Model Comparison Table
print('\n=== MODEL COMPARISON TABLE ===')
comparison_df = pd.DataFrame(results).set_index('Model')
print(comparison_df.round(4))

best_model_name = comparison_df['R² Score'].idxmax()
print(f'\n★ Best Model: {best_model_name} (R² = {comparison_df.loc[best_model_name, "R² Score"]:.4f})')

model_map = {
    'Linear Regression': lr_pipeline,
    'Decision Tree': dt_pipeline,
    'Random Forest': rf_pipeline,
}
best_pipeline = model_map[best_model_name]
```

## 6. Model Optimization

```python
# Optimize the best model
print(f'Optimizing: {best_model_name}')

if best_model_name in ['Random Forest', 'Decision Tree']:
    if best_model_name == 'Random Forest':
        param_distributions = {
            'regressor__n_estimators': [50, 100, 200, 300],
            'regressor__max_depth': [None, 10, 20, 30, 50],
            'regressor__min_samples_split': [2, 5, 10],
            'regressor__min_samples_leaf': [1, 2, 4],
            'regressor__max_features': ['sqrt', 'log2', None],
        }
        base_pipeline = rf_pipeline
        orig_metrics = rf_metrics
    else:
        param_distributions = {
            'regressor__max_depth': [None, 5, 10, 15, 20, 30],
            'regressor__min_samples_split': [2, 5, 10, 20],
            'regressor__min_samples_leaf': [1, 2, 4, 8],
            'regressor__max_features': ['sqrt', 'log2', None],
        }
        base_pipeline = dt_pipeline
        orig_metrics = dt_metrics
    
    search = RandomizedSearchCV(
        base_pipeline,
        param_distributions=param_distributions,
        n_iter=30,
        cv=5,
        scoring='r2',
        random_state=42,
        n_jobs=-1,
        verbose=1
    )
    print('Running RandomizedSearchCV (30 iterations, 5-fold CV)...')
    search.fit(X_train, y_train)
    
    print(f'\nBest Parameters:')
    for param, value in search.best_params_.items():
        print(f'  {param.replace("regressor__", "")}: {value}')
    print(f'Best CV R²: {search.best_score_:.4f}')
    
    opt_pred = search.predict(X_test)
    opt_metrics = evaluate_model(f'{best_model_name} (Optimized)', y_test, opt_pred)
    
    print(f'\nOriginal vs Optimized:')
    opt_comparison = pd.DataFrame([orig_metrics, opt_metrics]).set_index('Model')
    print(opt_comparison.round(4))
    
    if opt_metrics['R² Score'] > orig_metrics['R² Score']:
        print('\n✓ Optimized model is BETTER!')
        best_pipeline = search.best_estimator_
        final_metrics = opt_metrics
    else:
        print('\n✗ Optimization did not improve. Keeping original.')
        final_metrics = orig_metrics
else:
    print('Linear Regression - performing cross-validation...')
    cv_scores = cross_val_score(lr_pipeline, X_train, y_train, cv=5, scoring='r2')
    print(f'5-Fold CV R² Scores: {cv_scores.round(4)}')
    print(f'Mean CV R²: {cv_scores.mean():.4f} ± {cv_scores.std():.4f}')
    final_metrics = lr_metrics
```

## 7. Feature Importance

```python
# Feature Importance
cat_encoder = best_pipeline.named_steps['preprocessor'].transformers_[1][1]
try:
    cat_feature_names = cat_encoder.get_feature_names_out(feature_columns_categorical).tolist()
except:
    cat_feature_names = []
    for i, col in enumerate(feature_columns_categorical):
        cats = cat_encoder.categories_[i][1:]
        for cat in cats:
            cat_feature_names.append(f'{col}_{cat}')

all_feature_names = feature_columns_numeric + cat_feature_names

final_regressor = best_pipeline.named_steps['regressor']
if hasattr(final_regressor, 'feature_importances_'):
    importances = final_regressor.feature_importances_
    importance_df = pd.DataFrame({
        'Feature': all_feature_names,
        'Importance': importances
    }).sort_values('Importance', ascending=False)
    
    print('=== Top 15 Most Important Features ===')
    top_15 = importance_df.head(15).copy()
    top_15['Feature'] = top_15['Feature'].str.replace('Primary_Cuisine_', 'Cuisine: ')
    top_15['Feature'] = top_15['Feature'].str.replace('City_', 'City: ')
    top_15['Feature'] = top_15['Feature'].str.replace('_enc', '')
    print(top_15.to_string(index=False))
elif hasattr(final_regressor, 'coef_'):
    coef_df = pd.DataFrame({
        'Feature': all_feature_names,
        'Coefficient': final_regressor.coef_
    })
    coef_df['Abs_Coef'] = np.abs(coef_df['Coefficient'])
    print('=== Top 15 Features by Coefficient Magnitude ===')
    print(coef_df.sort_values('Abs_Coef', ascending=False).head(15)[['Feature', 'Coefficient']].to_string(index=False))
```

## 8. Customer Preference & Cuisine Analysis

```python
# Reload original data for cuisine analysis
df_cuisine = pd.read_csv('../dataset/restaurant.csv', encoding='latin-1')
df_cuisine['Cuisines'] = df_cuisine['Cuisines'].fillna('Unknown')

# Explode cuisines
cuisine_expanded = df_cuisine.assign(
    Cuisine=df_cuisine['Cuisines'].str.split(',')
).explode('Cuisine')
cuisine_expanded['Cuisine'] = cuisine_expanded['Cuisine'].str.strip()

print(f'Total unique cuisines: {cuisine_expanded["Cuisine"].nunique()}')
```

```python
# Cuisine statistics
cuisine_stats = cuisine_expanded.groupby('Cuisine').agg(
    Num_Restaurants=('Restaurant ID', 'nunique'),
    Total_Votes=('Votes', 'sum'),
    Avg_Rating=('Aggregate rating', 'mean')
).round(2)

# Top 10 by votes
print('=== Top 10 Most Popular Cuisines (by Total Votes) ===')
top_10_votes = cuisine_stats.sort_values('Total_Votes', ascending=False).head(10)
print(top_10_votes)
```

```python
# Highest-rated cuisines (with min 20 restaurants)
MIN_RESTAURANTS = 20
print(f'=== Highest-Rated Cuisines (min {MIN_RESTAURANTS} restaurants) ===')
popular = cuisine_stats[cuisine_stats['Num_Restaurants'] >= MIN_RESTAURANTS]
top_rated = popular.sort_values('Avg_Rating', ascending=False).head(15)
print(top_rated)
```

```python
# Customer Preference Comparison
print('=== Popular vs Highest-Rated ===')
top_popular = top_10_votes.index.tolist()
top_rated_list = top_rated.head(10).index.tolist()

print(f'{"Rank":<6} {"Most Popular (Votes)":<30} {"Highest Rated":<30}')
print('-' * 66)
for i in range(10):
    pop = top_popular[i] if i < len(top_popular) else '-'
    rated = top_rated_list[i] if i < len(top_rated_list) else '-'
    overlap = ' ★' if pop in top_rated_list[:10] else ''
    print(f'  {i+1:<4} {pop:<30} {rated:<30}{overlap}')

overlap_count = len(set(top_popular) & set(top_rated_list[:10]))
print(f'\nOverlap: {overlap_count} cuisines in BOTH lists (★)')

print('\n=== Key Findings ===')
print('1. Popularity (votes) does not equal highest rating')
print('2. Niche cuisines often have higher average ratings')
print('3. These are CORRELATIONS, not causal relationships')
```

## 9. Neural Network Model (TensorFlow/Keras)

```python
try:
    import tensorflow as tf
    from tensorflow import keras
    from tensorflow.keras import layers
    
    print(f'TensorFlow version: {tf.__version__}')
    
    # Preprocess for NN
    X_train_nn = best_pipeline.named_steps['preprocessor'].transform(X_train)
    X_test_nn = best_pipeline.named_steps['preprocessor'].transform(X_test)
    input_dim = X_train_nn.shape[1]
    
    # Build model
    nn_model = keras.Sequential([
        layers.Input(shape=(input_dim,)),
        layers.Dense(128, activation='relu'),
        layers.Dense(64, activation='relu'),
        layers.Dense(32, activation='relu'),
        layers.Dense(16, activation='relu'),
        layers.Dense(1, activation='linear')
    ])
    
    nn_model.compile(optimizer=keras.optimizers.Adam(0.001), loss='mse', metrics=['mae'])
    nn_model.summary()
    
    # Train
    history = nn_model.fit(X_train_nn, y_train, validation_split=0.2,
                           epochs=100, batch_size=64, verbose=1)
    
    # Evaluate
    nn_pred = nn_model.predict(X_test_nn, verbose=0).flatten()
    nn_metrics = evaluate_model('Neural Network', y_test, nn_pred)
    
    print('\n=== Neural Network Results ===')
    for k, v in nn_metrics.items():
        if k != 'Model': print(f'  {k}: {v:.4f}')
    
    # Complete comparison
    all_results = results + [nn_metrics]
    all_comparison = pd.DataFrame(all_results).set_index('Model')
    print('\n=== COMPLETE MODEL COMPARISON ===')
    print(all_comparison.round(4))
    
    best_overall = all_comparison['R² Score'].idxmax()
    print(f'\n★ Overall Best: {best_overall}')
    
except ImportError:
    print('TensorFlow not installed. Skipping neural network.')
except Exception as e:
    print(f'Error: {e}')
```

```python
# Prediction Function
def predict_restaurant_rating(input_data):
    """Predict Aggregate Rating for a restaurant."""
    input_processed = {
        'Average Cost for two': input_data.get('Average Cost for two', 500),
        'Longitude': input_data.get('Longitude', 0.0),
        'Latitude': input_data.get('Latitude', 0.0),
        'Votes': input_data.get('Votes', 0),
        'Price range': input_data.get('Price range', 2),
        'Num_Cuisines': input_data.get('Num_Cuisines', 1),
        'Has Table booking_enc': 1 if input_data.get('Has Table booking', 'No') == 'Yes' else 0,
        'Has Online delivery_enc': 1 if input_data.get('Has Online delivery', 'No') == 'Yes' else 0,
        'Is delivering now_enc': 1 if input_data.get('Is delivering now', 'No') == 'Yes' else 0,
        'City': input_data.get('City', 'Other'),
        'Primary_Cuisine': input_data.get('Primary_Cuisine', 'Other'),
    }
    input_df = pd.DataFrame([input_processed])
    input_df = input_df[feature_columns_numeric + feature_columns_categorical]
    prediction = best_pipeline.predict(input_df)[0]
    return round(np.clip(prediction, 0, 5), 2)

# Example predictions
example_1 = {
    'Average Cost for two': 2000, 'Longitude': 77.209, 'Latitude': 28.614,
    'Votes': 500, 'Price range': 4, 'Num_Cuisines': 3,
    'Has Table booking': 'Yes', 'Has Online delivery': 'No',
    'Is delivering now': 'No', 'City': 'New Delhi', 'Primary_Cuisine': 'North Indian',
}
print(f'Example 1 (Upscale North Indian, Delhi): {predict_restaurant_rating(example_1)} / 5.0')

example_2 = {
    'Average Cost for two': 300, 'Longitude': 77.209, 'Latitude': 28.614,
    'Votes': 50, 'Price range': 1, 'Num_Cuisines': 1,
    'Has Table booking': 'No', 'Has Online delivery': 'Yes',
    'Is delivering now': 'Yes', 'City': 'New Delhi', 'Primary_Cuisine': 'Fast Food',
}
print(f'Example 2 (Budget Fast Food, Delhi):     {predict_restaurant_rating(example_2)} / 5.0')
```
